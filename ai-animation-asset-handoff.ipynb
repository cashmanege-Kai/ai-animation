# AI Animation Asset Handoff
Kaggle version 1: https://www.kaggle.com/code/kaitsung/ai-animation-asset-handoff
Attach the input assets on Kaggle and save a run. Download asset-manifest.json from the saved version and use its paths and SHA-256 values to verify the assets before rendering.

```python
# AI Animation Asset Handoff
# Put approved animation assets under one attached Kaggle input folder.
from pathlib import Path
import hashlib
import json

input_root = Path('/kaggle/input')
output = Path('/kaggle/working/asset-manifest.json')
media_extensions = {'.png', '.jpg', '.jpeg', '.webp', '.mp4', '.mov', '.wav', '.mp3', '.srt', '.vtt'}
assets = []
for path in sorted(input_root.rglob('*')):
    if path.is_file() and path.suffix.lower() in media_extensions:
        digest = hashlib.sha256()
        with path.open('rb') as stream:
            for chunk in iter(lambda: stream.read(1024 * 1024), b''):
                digest.update(chunk)
        assets.append({'path': path.relative_to(input_root).as_posix(), 'bytes': path.stat().st_size, 'sha256': digest.hexdigest()})

output.parent.mkdir(parents=True, exist_ok=True)
output.write_text(json.dumps({'assets': assets}, ensure_ascii=False, indent=2), encoding='utf-8')
print(f'{len(assets)} media files listed in {output}')
```
